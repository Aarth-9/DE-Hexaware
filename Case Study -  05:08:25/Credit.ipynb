## Importing Libraries

```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Credit Card').getOrCreate()

df = spark.read.option("header",True).csv("credit card.csv")
df.printSchema()
```

### Credit card users in Spain

```python
from pyspark.sql.functions import col
df.filter(col("Geography") == "Spain").count()
```

```python
df.filter(col("Geography") == "Spain").show()
```

### Number of members who are elgible and active in the bank

```python
df.filter(
    (col("CreditScore") >= 700) &
    (col("Balance") > 0.0) &
    (col("IsActiveMember") == 1)
).count()
```
